# Cleaning Census / Poverty Indicators

This notebook loads the 2008-2012 City of Chicago Census socioeconomic-indicators file, normalizes the community-area names, subsets the six study neighborhoods (Austin, Humboldt Park, North Lawndale, Englewood, West Englewood, South Shore), and exports a poverty summary table to `docs/Pov_data_table.docx`.

*Note: this is a data-cleaning notebook. The canonical analysis and all published figures live in the `scripts/` directory.*

```python
import pandas as pd
from docx import Document
```

## Load the Census socioeconomic indicators

The raw file contains selected socioeconomic indicators for every Chicago community area, including the percent of households below the poverty line.

```python
poverty = pd.read_csv("../data/raw/Census_Data_-_Selected_socioeconomic_indicators_in_Chicago__2008___2012_20250408.csv")  # e.g., Poverty data 
```

## Subset the six study neighborhoods and export the table

Community-area names are title-cased so they match across datasets, the frame is filtered to the six highest-violence study neighborhoods, and the resulting poverty summary is written to a Word table at `docs/Pov_data_table.docx`.

```python
poverty.loc[:, "Community"] = poverty["COMMUNITY AREA NAME"].str.title()
selected_neighborhoods = ['Austin', 'Humboldt Park', 'North Lawndale', 'Englewood', 'West Englewood', 'South Shore']
poverty_six_hoods = poverty[poverty['COMMUNITY AREA NAME'].str.title().isin(selected_neighborhoods)].copy()
poverty_six_hoods = pd.DataFrame(poverty_six_hoods)

def dataframe_to_word(df, filename):
    doc = Document()
    table = doc.add_table(rows=1, cols=len(df.columns))

    hdr_cells = table.rows[0].cells
    for idx, col_name in enumerate(df.columns):
        hdr_cells[idx].text = str(col_name)

    for _, row in df.iterrows():
        row_cells = table.add_row().cells
        for idx, item in enumerate(row):
            row_cells[idx].text = str(item)

    doc.save(filename)

dataframe_to_word(poverty_six_hoods, "../docs/Pov_data_table.docx")
```
